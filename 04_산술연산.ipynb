```python
# 시리즈 연산

# 시리즈 객체 + 연산자(+,-,*,/) + 숫자(스칼라)

dict_data={
    "국어" : 100,
    "영어" : 80,
    "수학" : 90,
}

import pandas as pd

student1 = pd.Series(dict_data)
student1
```

```python
percentage  = student1 / 100
print(percentage)
```

```python
# 시리즈 vs 시리즈

student2 = pd.Series({"수학":80,"국어":90,"영어":80})

print(student1)
print( )
print(student2)
```

```python
addition =student1 +student2
addition
```

```python
subtraction = student1 - student2
subtraction
```

```python
multiplication = student1 * student2

multiplication
```

```python
divistion = student1 / student2
divistion
```

```python
result = pd.DataFrame([addition,subtraction,multiplication,divistion], 
            index =["덧셈",'뺄셈',"곱셈","나눗셈"] )

result

```

```python
# Nan 값이 있는 시리즈 연산
import pandas as pd
import numpy as np

student1 = pd.Series({
    "국어" : np.nan,
    "영어": 80 , 
    "수학" : 90
})

student2 = pd. Series({
    "수학": 80,
    "국어":90
})

print(student1)
print()
print(student2)
```

```python

addition = student1 + student2
substraction = student1 - student2
multiplication = student1 * student2
divistion = student1 / student2

print(addition)
print(subtraction)
print(multiplication)
print(divistion)
```

```python
result = pd.DataFrame([
    addition,subtraction,multiplication,divistion
],index=["덧셈",'뺄셈','곱셈','나눗셈'])

result
```
