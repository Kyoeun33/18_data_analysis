```python
import requests
import pandas as pd

my_keys ="--카카오 키 --"

def get_geocoding(place):
    url = f"http://dapi.kakao.com/v2/local/search/address.json?query={place}"
    headers = {"Authorization" : f" KakaoAK{my_key}"}
    response = requests.get(url,headers = headers)
    result = response . json()
    return result["document"][0]["y"],result["documents"][0]["x"]

lat = []
lng = []

places = [
    "서울특별시 종로구 세종대로 175",
    "서울 특별시 서초구 처초동 700",
    "부산 광역시 해운대구 해운대해변로 364"
]

i =  0

for place in places :
    i = i+1
    try :
        print(i,place)
        place_lat,place_lon=get_geocoding(place)
        lng.append(place_lat)
        lng.append(place_lon)

    except : 
        lat.append("")
        lng.append("")
        print(i,"error")


df =pd.DataFrame({
    "위도": lat,
    "경도": lng
}, index = places)

df
```
