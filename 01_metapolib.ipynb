```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_excel("./data/시도별_전출입_인구수.xlsx")
df.head()
```

```python
df = df.ffill() # 누락 값을 앞 데이터로 메움
df.head()
```

```python
df.info()
```

```python
mask = (df["전출지별"] =="서울특별시") & (df["전입지별"] != "서울특별시")
df_seoul=df[mask]
df_seoul
```

```python
df_seoul = df_seoul.drop(["전출지별"],axis=1)
df_seoul
```

```python
df_seoul = df_seoul.rename({'전입지별':"전입지"},axis = 1)
df_seoul = df_seoul.set_index("전입지")

df_seoul
```

```python
df_seoul = df_seoul.set_index("전입지")

df_seoul
```

```python
sr_one = df_seoul.loc["경기도"]
plt.plot(sr_one.index,sr_one.values)

plt.show()
```

```python
plt.plot(sr_one)
plt.show()
```

```python
# 차트 제목 / 축 이름 / 선 스타일

plt.plot(sr_one.index,sr_one.values,linestyle="--")

plt.title("서울 ==> 경기 인구 이동")

plt.xlabel("기간")
plt.show()
```

```python
from matplotlib import font_manager, rc
font_path="./data/malgun.ttf"
font_name = font_manager.FontProperties(fname=font_path).get_name()
rc('font',family = font_name)
plt.plot(sr_one.index,sr_one.values,linestyle="--")

plt.title("서울 ==> 경기 인구 이동")

plt.xlabel("기간")
plt.show()
```

```python

sr1 = df_seoul.loc["경기도"]

plt.figure(figsize = (14,5))

plt.xticks(rotation="vertical")
plt.plot(sr1.index, sr1.values)

plt.plot(sr_one.index,sr_one.values,linestyle="--")

plt.title("서울 ==> 경기 인구 이동")

plt.xlabel("기간")
plt.show()
```

```python


plt.style.use("ggplot")

plt.figure(figsize=(14,5))

plt.xticks(rotation="vertical")
plt.plot(sr1.index, sr1.values,
         marker = "o",
         markerfacecolor = "red",
         markeredgewidth = 2,
         markersize = 10)

plt.plot(sr1.index,sr_one.values,linestyle="--")

plt.title("서울 ==> 경기 인구 이동",size = 30)

plt.xlabel("기간",size =20)
plt.ylabel("인구이동수",size =20)

plt.legend(labels=["서울 -> 경기"],loc='best', fontsize = 15)

plt.show()
```

```python
#figure 객체, Axes 객체

import pandas as pd
import matplotlib.pyplot as plt

samsung_revenue = pd.read_csv("data/삼성전자_분기별_매출액.csv")
# samsung_revenue.sort_values
samsung_revenue.sort_values
samsung_revenue.head()
```

```python
fig = plt.figure()
fig
```

```python
fig,axe = plt.subplots()

plt.subplots()
```

```python
print(fig)
print(axe)
```

```python
fig, axe = plt.subplots(figsize = (8,2))
axe.plot(samsung_revenue["quarter"], samsung_revenue["value"])
plt.show()
```

```python
fig,aces = plt.subplots(1,2)

print(fig)
```

```python
fig , axes = plt.subplots(1,2,figsize=(12,2))
axes[0].plot(samsung_revenue["quarter"],samsung_revenue["value"])
axes[1].plot(samsung_revenue["quarter"],samsung_revenue["value"])
```

```python
fig , axes = plt.subplots(1,2,figsize=(12,2))
axes[0].plot(samsung_revenue["quarter"],samsung_revenue["value"])
samsung_revenue["value"].plot(ax = axes[1])
```

```python
fig,axes = plt.subplots(2,2)
print(fig)
print(axes)
```

```python
fig , axes = plt.subplots(2,2, figsize = (12,2), constrained_layout=True)
axes [0,1],plot(samsung_revenue['quarter'],samsung_revenue['value'])
plt.show()
```

```python
fig,axe = plt.subplot_mosaic(
    [["top_left","right"],
    ["botten_left ","right"]]
)

print(fig)
print(axe)
```

```python
fig,axe = plt.subplot_mosaic(
    [["top_left","right"],
    ["botten_left ","right"]],figsize = (12,4)
)

axe["right"].plot(samsung_revenue["quarter"],samsung_revenue["value"])
plt.show()
```
