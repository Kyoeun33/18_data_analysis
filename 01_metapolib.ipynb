```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_excel("./data/시도별_전출입_인구수.xlsx")
df.head()
```

```python
df = df.ffill() # 누락 값을 앞 데이터로 메움
df.head()
```

```python
df.info()
```

```python
mask = (df["전출지별"] =="서울특별시") & (df["전입지별"] != "서울특별시")
df_seoul=df[mask]
df_seoul
```

```python
df_seoul = df_seoul.drop(["전출지별"],axis=1)
df_seoul
```

```python
df_seoul = df_seoul.rename({'전입지별':"전입지"},axis = 1)
df_seoul = df_seoul.set_index("전입지")

df_seoul
```

```python
df_seoul = df_seoul.set_index("전입지")

df_seoul
```

```python
sr_one = df_seoul.loc["경기도"]
plt.plot(sr_one.index,sr_one.values)

plt.show()
```

```python
col_years = list(map(str, range(1970,2018)))
df_3 = df_seoul.loc[["충청남도","경상북도","강원도"],col_years]
df_3

plt.style.use("ggplot")

fig = plt.figure(figsize=(20,5))
ax = fig.add_subplot(1,1,1)


ax.plot(col_years, df_3.loc["충청남도",:],
        marker="o",
        markerfacecolor = "green",
        markersize = 10,
        color = "olive",
        linewidth = 2,
        label = "서울 -> 충청"
        )

ax.plot(col_years, df_3.loc["경상북도",:],
        marker="o",
        markerfacecolor = "skyblue",
        markersize = 10,
        linewidth = 2,
        color = "skyblue",
        label = "서울 -> 경상"
        )

ax.plot(col_years, df_3.loc["강원도",:],
        marker="o",
        markerfacecolor = "red",
        markersize = 10,
        linewidth = 2,
        color = "magenta",
        label = "서울 -> 강원"
        )


ax.legend(loc= "best")


ax.set_title("서울 -> 충남, 경북, 강원 인구 이동", size =20)

ax.set_xlabel ( "기간", size = 10)
ax.set_ylabel ( "이동 인구수", size = 10)

ax.set_xticks(range(len(col_years)))
ax.set_xticklabels(col_years,rotation = 90)

ax.tick_params(axis="x",labelsize=10)
ax.tick_params(axis="y",labelsize=10)

plt.show()
```

```python
col_years = list(map(str,range(1970,2018)))
df_4 = df_seoul.loc[["충청남도","경상북도","강원도","전라남도"], col_years]

print(df_4.dtypes)
df_4 = df_4.astype(int)
print(df_4.dtypes)
```

```python
df_4 = df_4.transpose()
df_4.head()
```

```python
plt.style.use("ggplot")

df_4.plot(
    kind="area",
    stacked=False,
    alpha=0.3,
    figsize=(20,10),

)

plt.title("서울==>타시도 인구 이동")
plt.ylabel("이동인구수")
plt.xlabel("기관")
plt.legend(loc="best")

plt.show()
```

```python
plt.style.use("ggplot")

df_4.plot(
    kind="area",
    stacked=True,
    alpha=0.3,
    figsize=(20,10),

)

plt.title("서울==>타시도 인구 이동")
plt.ylabel("이동인구수")
plt.xlabel("기관")
plt.legend(loc="best")

plt.show()
```

```python
# 막대그래프 그리기
col_years = list(map(str,range(2010,2018)))
df_4 = df_seoul.loc[["충청남도","경상북도","강원도","전라남도"], col_years]
df_4=df_4.T

df_4.index = df_4.index.map(int)

df_4.plot(
    kind="bar",
    figsize=(20,10),
    width=0.7,
    color = ["orange","blue","magenta","green"] 
)


plt.title("서울==>타시도 인구 이동")
plt.ylabel("이동인구수")
plt.xlabel("기관")
plt.ylim(5000,30000)
plt.legend(loc="best")

plt.show()
```
